# Prompt

---

Write a Python example to demonstrate hierarchical inheritance, where two child classes inherit from the same parent class and use super() to call the parent method.

```python
class A:
    def method(self):
        print("Method in class A")

class B(A):
    def method(self):
        super().method()  # Call method from class A
        print("Method in class B")

class C(A):
    def method(self):
        super().method()  # Call method from class A
        print("Method in class C")
```

# 1. Reverse an Array in groups


---
Given an array arr[ ] and an integer k, find the array after reversing every subarray of consecutive k elements in place. If the last subarray has fewer than k elements, reverse it as it is. Modify the array in place, do not return anything.


# PROMPT


---

Write a Python function to reverse an array in groups of k elements. If any elements are left at the end, reverse them too.

# Code


---



```python
def reverseInGroups(arr, k):
    i = 0
    n = len(arr)

    while i < n:
        left = i

        right = min(i + k - 1, n - 1)

        while left < right:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
            right -= 1

        i += k


if __name__ == "__main__":
    arr = [1, 2, 3, 4, 5, 6, 7, 8]
    k = 3
    reverseInGroups(arr, k)
    print(" ".join(map(str, arr)))
```

# 2. Duplicates in Limited Range and Limited Repetition Array


---

Given an array arr[ ] of integers of size n, where each element is in the range 1 to n and each element can occur at most twice, find all elements that appear twice in the array.


# Prompt


---
Write a Python function to find all the duplicate elements in an array and return them without repeating any element in the result


# Code

---



```python
def findDuplicates(arr):
    ans = []

    for i in range(len(arr)):
        cnt = 0

        for it in ans:
            if arr[i] == it:
                cnt += 1
                break

        if cnt:
            continue


        for j in range(i + 1, len(arr)):
            if arr[i] == arr[j]:
                cnt += 1
                break

        if cnt:
            ans.append(arr[i])

    return ans


if __name__ == "__main__":
    arr = [2, 3, 1, 2, 3]
    res = findDuplicates(arr)
    print(*res)
```

# 3. Array after k Rotations


---

Given an array arr[ ] and an integer k, rotate the array in place k times to the right (clockwise). In each rotation, the last element moves to the front, and all other elements shift one position to the right. Modify the array in place, do not return anything.

# Prompt


---

Write a Python function to rotate an array clockwise by k positions

# Code
---

```python
def rotateclockwise(arr, k):
    if k == 0:
        return

    n = len(arr)

    temp = arr[n - 1]
    for i in range(n - 1, 0, -1):
        arr[i] = arr[i - 1]
    arr[0] = temp

    rotateclockwise(arr, k - 1)

if __name__ == "__main__":
    arr = [1, 2, 3, 4, 5, 6]
    k = 2
    rotateclockwise(arr, k)
    for it in arr:
        print(it, end=" ")
```

# 4. Maximum Subarray Sum - Kadane's Algorithm


---

Given an integer array arr[ ], find the subarray (containing at least one element) which has the maximum possible sum, and return that sum. A subarray is a continuous part of an array.

# Prompt


---

Write a Python function to find the maximum sum of a contiguous subarray in a given array. Include an example with output.

# Code


---



```python
def maxSubarraySum(arr):
    res = arr[0]

    for i in range(len(arr)):
        currSum = 0

        for j in range(i, len(arr)):
            currSum = currSum + arr[j]

            res = max(res, currSum)

    return res

if __name__ == "__main__":
    arr = [2, 3, -8, 7, -1, 2, 3]
    print(maxSubarraySum(arr))
```

# 5. Majority Element


---
Given an array arr[] of size n, find the element that appears more than ⌊n/2⌋ times. If no such element exists, return -1.


# Prompt

---

Write a Python function to find the majority element in an array, which is the element that appears more than half the time. Return -1 if no such element exists. Include an example.

# Code


---



```python
def majorityElement(arr):

    n = len(arr)

    for i in range(n):
        count = 0

        for j in range(n):
            if arr[i] == arr[j]:
                count += 1

        if count > n // 2:
            return arr[i]

    return -1

if __name__ == "__main__":
	arr = [1, 1, 2, 1, 3, 5, 1]
	print(majorityElement(arr))
```

# Prompt

---

Write a Python function to check whether an array is sorted in ascending order. Return True if it is sorted, otherwise return False. Include an example.

# Code

---



```python
def isSorted(arr):

    n = len(arr)

    for i in range(1, n):
        if (arr[i-1] > arr[i]):
            return False

    return True

if __name__ == "__main__":
  arr = [10, 20, 30, 40, 50]
  n = len(arr)
  if (isSorted(arr)):
     print("true")
  else:
     print("false")
```

---

Write a Python function to search for an element in an array and return its index if found, otherwise return -1.

# 9. Count substrings with k distinct characters

---
Given a string s consisting of only lowercase English letters and an integer k, count the total number of substrings (not necessarily distinct) of s that contain exactly k distinct characters.


# Prompt


---

Write a Python function to count the number of substrings that contain exactly k distinct characters. Include an example with output.

# Code

---



```python
def countSubstr(s, k):
    n = len(s)
    ans = 0

    for i in range(n):


        map = [False] * 26
        distinctCnt = 0

        for j in range(i, n):


            if not map[ord(s[j]) - ord('a')]:
                map[ord(s[j]) - ord('a')] = True
                distinctCnt += 1

            if distinctCnt == k:
                ans += 1

    return ans

if __name__ == "__main__":
    s = "abc"
    k = 2
    print(countSubstr(s, k))
```

# Program for Square Root of Integer


---
Given a positive integer n, find its square root. If n is not a perfect square, then return floor of √n.


# Prompt

---

Write a Python function to find the floor of the square root of a given number n

# Code

---



```python
def floorSqrt(n):

    res = 1
    while res * res <= n:
        res += 1

    return res - 1

if __name__ == "__main__":
    n = 11
    print(floorSqrt(n))
```
