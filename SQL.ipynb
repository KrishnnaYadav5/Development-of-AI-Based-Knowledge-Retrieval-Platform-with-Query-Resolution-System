UPDATE

---
Write a query to update the marks of the student whose ID is 2 to 95 and display the updated Students table ?


Prompt: Write a SQL using pyspark to update the marks of the student whose ID is 2 to 95 and display the updated Students table.

```python
from pyspark.sql import functions as F

# 1. Load the table into a DataFrame
df = spark.read.table("Students")

# 2. Update the marks column where id = 2 using F.when
updated_df = df.withColumn("marks", F.when(df["id"] == 2, 95).otherwise(df["marks"]))

# 3. Break the lineage to prevent self-overwrite exceptions
local_data = updated_df.collect()
schema = updated_df.schema
final_df = spark.createDataFrame(local_data, schema)

# 4. Overwrite the table with the modified dataset
final_df.write.format("parquet").mode("overwrite").saveAsTable("Students")

# 5. Show the updated table contents
spark.sql("SELECT * FROM Students").show()
```

DELETE

---
Write a query to delete the student record whose ID is 3 from the Students table and display the remaining records.


Prompt: Write a SQL using pyspark to delete the student record whose ID is 3 from the Students table and display the remaining records.

```python
# 1. Load the table into a DataFrame
df = spark.read.table("Students")

# 2. Filter out the record we want to delete (keeping id != 3)
filtered_df = df.filter(df["id"] != 3)

# 3. Break the lineage to prevent self-overwrite exceptions
local_data = filtered_df.collect()
schema = filtered_df.schema
final_df = spark.createDataFrame(local_data, schema)

# 4. Overwrite the table with the remaining records
final_df.write.format("parquet").mode("overwrite").saveAsTable("Students")

# 5. Show the updated table contents
spark.sql("SELECT * FROM Students").show()
```

GROUP BY

---

Write a query to calculate the average marks of students in each city ?

Prompt: Write a SQL using pyspark to calculate the average marks of students in each city and display the results.

```python
spark.sql("""
    SELECT city, AVG(marks) AS average_marks
    FROM Students
    GROUP BY city
""").show()
```
